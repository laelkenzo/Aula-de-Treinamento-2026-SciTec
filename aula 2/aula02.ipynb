```python
import pandas as pd

# 1. Carregando o arquivo original
df = pd.read_csv("ecommerce_customer_behavior_dataset.csv")

# 2. Diagnóstico Inicial (O seu ponto de partida)
print("=== PESO INICIAL DO DATASET ===")
df.info(memory_usage="deep")
print("\n" + "=" * 60 + "\n")


# 3. APLICANDO A OTIMIZAÇÃO MÁXIMA (Coluna por Coluna)

colunas_categoricas = [
    "Gender",
    "City",
    "Membership_Type",
    "Satisfaction_Level",
    "Device_Type",
]
for col in colunas_categoricas:
    if col in df.columns:
        df[col] = df[col].astype("category")

colunas_booleanas = ["Is_Returning_Customer"]
for col in colunas_booleanas:
    if col in df.columns:
        df[col] = df[col].astype("bool")

for col in df.select_dtypes(include=["int64"]).columns:
    df[col] = pd.to_numeric(df[col], downcast="integer")

for col in df.select_dtypes(include=["float64"]).columns:
    df[col] = pd.to_numeric(df[col], downcast="float")


print("=== PESO APÓS OTIMIZAÇÃO MÁXIMA ===")
df.info(memory_usage="deep")
```
