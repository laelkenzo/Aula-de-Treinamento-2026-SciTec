```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Carrega os dados do arquivo que você colocou na pasta
df = pd.read_csv("ecommerce_customer_behavior_dataset.csv")

# 2. Conta quantas vezes cada gênero aparece nas 5.000 linhas
contagem_genero = df["Gender"].value_counts()

# 3. Manda o Python desenhar o gráfico de pizza com as porcentagens
contagem_genero.plot(kind="pie", autopct="%1.1f%%")

# 4. Coloca um título no gráfico
plt.title("Distribuição de Clientes por Gênero")

# 5. Remove uma legenda automática na lateral para o gráfico ficar limpo
plt.ylabel("")

# 6. Exibe o gráfico finalizado na tela
plt.show()
```

```python
# Cria o gráfico de histograma para a coluna de idade (Age)
df["Age"].plot(kind="hist", bins=10, edgecolor="black")

# Adiciona os textos
plt.title("Distribuição de Idade dos Clientes")
plt.xlabel("Idade")
plt.ylabel("Quantidade de Clientes")

# Mostra o gráfico
plt.show()
```

```python
# Mostra os nomes reais de todas as colunas do seu arquivo
df.columns
```

```python
# 1. Agrupa os clientes por cidade e calcula a média do valor gasto (Total_Amount)
gasto_por_cidade = df.groupby("City")["Total_Amount"].mean()

# 2. Cria o gráfico de colunas/barras verticais
gasto_por_cidade.plot(kind="bar")

# 3. Adiciona as etiquetas de texto para ficar fácil de ler
plt.title("Média de Gasto por Cidade")
plt.xlabel("Cidades")
plt.ylabel("Gasto Médio ($)")

# 4. Mostra o gráfico na tela
plt.show()
```

```python
# Cria um gráfico do tipo Boxplot relacionando o método de pagamento com o valor total da compra
df.boxplot(column="Total_Amount", by="Payment_Method")

# Adiciona os títulos e etiquetas de texto
plt.title("Distribuição do Valor de Compra por Método de Pagamento")
plt.suptitle("")  # Limpa um título automático chato que o pandas coloca no topo
plt.xlabel("Método de Pagamento")
plt.ylabel("Valor da Compra ($)")

# Rotaciona os nomes dos métodos de pagamento para eles não ficarem embolados
plt.xticks(rotation=45)

# Mostra o gráfico na tela
plt.show()
```
