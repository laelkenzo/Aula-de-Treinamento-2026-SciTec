```python
valores_brutos = [10, 50, 150, 20, 300]

def processar_alertas(valores: list[int]) -> list[int]:
    alertas = []
    for valor in valores:
        if valor > 100:
            alertas.append(valor)
    return alertas

alertas_gerados = processar_alertas(valores_brutos)

if not alertas_gerados:
    print("Sistema Seguro")
else:
    print(f"⚠️ Revisar valores: {alertas_gerados}")
```
